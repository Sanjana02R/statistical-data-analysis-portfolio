## Objective

The objective of this study is to analyze the relationship between IPL player performance characteristics and their auction prices using statistical analysis and regression models.

The analysis specifically examines whether batting performance indicators such as **batting strike rate** and **number of sixes** are associated with a player's **IPL auction sold price**.

## Research Questions

1. Is there a statistically significant relationship between batting strike rate and IPL player sold price?

2. What is the regression relationship between batting strike rate and sold price?

3. How much variation in player sold price is explained by batting strike rate?

4. Does the number of sixes improve the prediction of IPL player auction price?

5. How does a multiple regression model using batting strike rate and number of sixes compare with a simple regression model?

6. What is the estimated impact of batting strike rate and number of sixes on a player's sold price?



```python
# ============================================================
# IPL PLAYER AUCTION PRICE ANALYSIS
# DATA SOURCE: IPL RAW DATA SHEET
# ============================================================


# ------------------------------------------------------------
# INSTALL REQUIRED PACKAGE FOR .XLS FILES
# ------------------------------------------------------------

!pip -q install xlrd


# ------------------------------------------------------------
# IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import statsmodels.api as sm
import statsmodels.formula.api as smf

from scipy import stats


# ------------------------------------------------------------
# LOAD THE RAW IPL DATA
# ------------------------------------------------------------

# IMPORTANT:
# This code assumes you have already uploaded:
# Chapter 10 IPL CASE DATA.xls

file_name = "Chapter 10 IPL CASE DATA.xls"


# Read ONLY the raw data sheet
df = pd.read_excel(
    file_name,
    sheet_name="IPL Raw Data"
)


print("=" * 70)
print("IPL PLAYER AUCTION PRICE ANALYSIS")
print("=" * 70)

print("\nDataset successfully loaded!")

print("\nDataset dimensions:")
print(df.shape)

print("\nNumber of players:")
print(len(df))

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five observations:")
display(df.head())


# ------------------------------------------------------------
# DATASET INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())


# ------------------------------------------------------------
# SELECT VARIABLES REQUIRED FOR ANALYSIS
# ------------------------------------------------------------

# The actual raw dataset columns contain spaces.
# We select the relevant variables and rename them
# for easier Python analysis.

analysis_df = df[
    [
        "PLAYER NAME",
        "SR -B",
        "SIXERS",
        "SOLD PRICE"
    ]
].copy()


# Rename variables

analysis_df.columns = [

    "Player_Name",

    "Batting_Strike_Rate",

    "Sixes",

    "Sold_Price"

]


# ------------------------------------------------------------
# CLEAN DATA
# ------------------------------------------------------------

analysis_df = analysis_df.replace(
    [np.inf, -np.inf],
    np.nan
)


analysis_df = analysis_df.dropna()


# Convert variables to numeric

analysis_df["Batting_Strike_Rate"] = pd.to_numeric(
    analysis_df["Batting_Strike_Rate"],
    errors="coerce"
)


analysis_df["Sixes"] = pd.to_numeric(
    analysis_df["Sixes"],
    errors="coerce"
)


analysis_df["Sold_Price"] = pd.to_numeric(
    analysis_df["Sold_Price"],
    errors="coerce"
)


analysis_df = analysis_df.dropna()


print("\n" + "=" * 70)
print("CLEAN DATASET")
print("=" * 70)

print("\nNumber of observations used:")
print(len(analysis_df))

display(analysis_df.head())


# ------------------------------------------------------------
# DESCRIPTIVE STATISTICS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

display(
    analysis_df[
        [
            "Batting_Strike_Rate",
            "Sixes",
            "Sold_Price"
        ]
    ].describe()
)


# ============================================================
# EXPLORATORY DATA ANALYSIS
# ============================================================


# ------------------------------------------------------------
# SCATTER PLOT:
# BATTING STRIKE RATE VS SOLD PRICE
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=analysis_df,
    x="Batting_Strike_Rate",
    y="Sold_Price"
)

plt.title(
    "Batting Strike Rate vs IPL Player Sold Price"
)

plt.xlabel(
    "Batting Strike Rate"
)

plt.ylabel(
    "Sold Price (US$)"
)

plt.show()


# ------------------------------------------------------------
# SCATTER PLOT:
# SIXES VS SOLD PRICE
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=analysis_df,
    x="Sixes",
    y="Sold_Price"
)

plt.title(
    "Number of Sixes vs IPL Player Sold Price"
)

plt.xlabel(
    "Number of Sixes"
)

plt.ylabel(
    "Sold Price (US$)"
)

plt.show()


# ------------------------------------------------------------
# SOLD PRICE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    analysis_df["Sold_Price"],
    kde=True
)

plt.title(
    "Distribution of IPL Player Sold Prices"
)

plt.xlabel(
    "Sold Price (US$)"
)

plt.show()


# ------------------------------------------------------------
# CORRELATION MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CORRELATION ANALYSIS")
print("=" * 70)


correlation_data = analysis_df[
    [
        "Batting_Strike_Rate",
        "Sixes",
        "Sold_Price"
    ]
]


print("\nCorrelation Matrix:")

display(
    correlation_data.corr()
)


plt.figure(figsize=(6, 4))

sns.heatmap(
    correlation_data.corr(),
    annot=True,
    fmt=".3f"
)

plt.title(
    "Correlation Between Performance Variables and Sold Price"
)

plt.show()


# ============================================================
# QUESTION 1
# RELATIONSHIP BETWEEN BATTING STRIKE RATE AND SOLD PRICE
# ============================================================

print("\n" + "=" * 70)
print("QUESTION 1")
print("RELATIONSHIP BETWEEN BATTING STRIKE RATE AND SOLD PRICE")
print("=" * 70)


correlation, p_value = stats.pearsonr(

    analysis_df["Batting_Strike_Rate"],

    analysis_df["Sold_Price"]

)


print(
    f"\nCorrelation Coefficient: "
    f"{correlation:.4f}"
)


print(
    f"P-value: "
    f"{p_value:.4f}"
)


print("\nHypotheses:")

print(
    "H0: Batting strike rate has no relationship "
    "with IPL player sold price."
)

print(
    "H1: Batting strike rate is related "
    "to IPL player sold price."
)


if p_value < 0.05:

    print(
        "\nConclusion:"
    )

    print(
        "Reject H0. There is statistically significant "
        "evidence of a relationship between batting strike "
        "rate and IPL player sold price."
    )


else:

    print(
        "\nConclusion:"
    )

    print(
        "Fail to reject H0. There is insufficient "
        "statistical evidence of a significant relationship "
        "between batting strike rate and IPL player sold price."
    )


# ============================================================
# QUESTION 2
# SIMPLE LINEAR REGRESSION
# ============================================================

print("\n" + "=" * 70)
print("QUESTION 2")
print("SIMPLE LINEAR REGRESSION")
print("=" * 70)


# Model:
#
# Sold Price = β0 + β1(Batting Strike Rate)


simple_model = smf.ols(

    "Sold_Price ~ Batting_Strike_Rate",

    data=analysis_df

).fit()


print(
    simple_model.summary()
)


# Extract coefficients

intercept = simple_model.params["Intercept"]

beta_strike_rate = simple_model.params[
    "Batting_Strike_Rate"
]


print("\nRegression Equation:")

print(
    f"Sold Price = "
    f"{intercept:.2f} + "
    f"({beta_strike_rate:.2f} × Batting Strike Rate)"
)


# ============================================================
# QUESTION 3
# R-SQUARED INTERPRETATION
# ============================================================

print("\n" + "=" * 70)
print("QUESTION 3")
print("MODEL EXPLANATORY POWER")
print("=" * 70)


r_squared = simple_model.rsquared


print(
    f"\nR-squared: "
    f"{r_squared:.4f}"
)


print(
    f"\nInterpretation:"
)


print(
    f"Approximately "
    f"{r_squared * 100:.2f}% "
    f"of the variation in IPL player sold price "
    f"is explained by batting strike rate."
)


# ------------------------------------------------------------
# SIGNIFICANCE OF STRIKE RATE
# ------------------------------------------------------------

strike_rate_pvalue = simple_model.pvalues[
    "Batting_Strike_Rate"
]


print(
    f"\nP-value for Batting Strike Rate: "
    f"{strike_rate_pvalue:.4f}"
)


if strike_rate_pvalue < 0.05:

    print(
        "\nThe batting strike rate coefficient is "
        "statistically significant at the 5% level."
    )


else:

    print(
        "\nThe batting strike rate coefficient is not "
        "statistically significant at the 5% level."
    )


# ============================================================
# QUESTION 4
# MULTIPLE LINEAR REGRESSION
# ============================================================

print("\n" + "=" * 70)
print("QUESTION 4")
print("MULTIPLE LINEAR REGRESSION")
print("=" * 70)


# Model:
#
# Sold Price =
#
# β0 +
#
# β1(Batting Strike Rate) +
#
# β2(Number of Sixes)


multiple_model = smf.ols(

    "Sold_Price ~ Batting_Strike_Rate + Sixes",

    data=analysis_df

).fit()


print(
    multiple_model.summary()
)


# Extract coefficients

multiple_intercept = multiple_model.params[
    "Intercept"
]


multiple_beta_sr = multiple_model.params[
    "Batting_Strike_Rate"
]


multiple_beta_sixes = multiple_model.params[
    "Sixes"
]


print("\nMultiple Regression Equation:")


print(

    f"Sold Price = "

    f"{multiple_intercept:.2f} "

    f"+ ({multiple_beta_sr:.2f} × Batting Strike Rate) "

    f"+ ({multiple_beta_sixes:.2f} × Number of Sixes)"

)


# ============================================================
# QUESTION 5
# MODEL COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("QUESTION 5")
print("COMPARE SIMPLE AND MULTIPLE REGRESSION MODELS")
print("=" * 70)


comparison = pd.DataFrame({

    "Model": [

        "Simple Regression",

        "Multiple Regression"

    ],

    "R-Squared": [

        simple_model.rsquared,

        multiple_model.rsquared

    ],

    "Adjusted R-Squared": [

        simple_model.rsquared_adj,

        multiple_model.rsquared_adj

    ],

    "AIC": [

        simple_model.aic,

        multiple_model.aic

    ],

    "BIC": [

        simple_model.bic,

        multiple_model.bic

    ]

})


display(
    comparison
)


print("\nInterpretation:")


if multiple_model.rsquared_adj > simple_model.rsquared_adj:

    print(

        "The multiple regression model has a higher "
        "Adjusted R-squared value. Therefore, adding "
        "the number of sixes improves the explanatory "
        "power of the regression model."

    )


else:

    print(

        "Adding the number of sixes does not produce "
        "a meaningful improvement in the model's "
        "Adjusted R-squared value."

    )


# ============================================================
# QUESTION 6
# INTERPRETATION OF REGRESSION COEFFICIENTS
# ============================================================

print("\n" + "=" * 70)
print("QUESTION 6")
print("INTERPRETATION OF COEFFICIENTS")
print("=" * 70)


print(

    f"\nBatting Strike Rate Coefficient: "

    f"{multiple_beta_sr:,.2f}"

)


print(

    f"\nInterpretation: Holding the number of sixes "
    f"constant, a one-unit increase in batting strike "
    f"rate is associated with an estimated change of "
    f"${multiple_beta_sr:,.2f} in player sold price."

)


print(

    f"\nNumber of Sixes Coefficient: "

    f"{multiple_beta_sixes:,.2f}"

)


print(

    f"\nInterpretation: Holding batting strike rate "
    f"constant, scoring one additional six is associated "
    f"with an estimated change of "
    f"${multiple_beta_sixes:,.2f} in player sold price."

)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)


print(

    "\nThis project analyzed IPL player auction prices "
    "using data from 130 IPL players."

)


print(

    "\nThe analysis included:"
)


print(

    "1. Data exploration"
)


print(

    "2. Descriptive statistics"
)


print(

    "3. Correlation analysis"
)


print(

    "4. Simple linear regression"
)


print(

    "5. Multiple linear regression"
)


print(

    "6. Model comparison"
)


print(

    "7. Interpretation of regression coefficients"
)


print(

    "\nThe results demonstrate how player performance "
    "characteristics can be statistically analyzed to "
    "understand their relationship with IPL auction prices."

)
```
